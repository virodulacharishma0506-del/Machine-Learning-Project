```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
total_runs = df["total_runs"].sum()

print("Total Runs:", total_runs)
```

```python
batsman_runs = df["batsman_runs"].sum()

print("Total Batsman Runs:", batsman_runs)
```

```python
extra_runs = df["extra_runs"].sum()

print("Total Extra Runs:", extra_runs)
```

```python
team_runs = df.groupby("batting_team")["total_runs"].sum()

print(team_runs.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))

team_runs.sort_values(ascending=False).plot(kind="bar")

plt.title("Total Runs by Batting Team")
plt.xlabel("Batting Team")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)

plt.show()
```

```python
top_batters = (
    df.groupby("batter")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_batters)
```

```python
plt.figure(figsize=(10,6))

top_batters.plot(kind="bar")

plt.title("Top 10 Batters by Runs")
plt.xlabel("Batter")
plt.ylabel("Runs")
plt.xticks(rotation=45)

plt.show()
```

```python
total_wickets = df["is_wicket"].sum()

print("Total Wickets:", total_wickets)
```

```python
dismissal_types = df["dismissal_kind"].value_counts()

print(dismissal_types)
```

```python
plt.figure(figsize=(10,6))

dismissal_types.plot(kind="bar")

plt.title("Dismissals by Type")
plt.xlabel("Dismissal Type")
plt.ylabel("Number of Wickets")
plt.xticks(rotation=45)

plt.show()
```

```python
top_bowlers = (
    df[df["is_wicket"] == 1]
    .groupby("bowler")["is_wicket"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_bowlers)
```

```python
plt.figure(figsize=(10,6))

top_bowlers.plot(kind="bar")

plt.title("Top 10 Bowlers by Wickets")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="total_runs")

plt.title("Distribution of Runs per Delivery")
plt.xlabel("Runs per Delivery")
plt.ylabel("Number of Deliveries")

plt.show()
```

```python
runs_by_over = df.groupby("over")["total_runs"].sum()

print(runs_by_over)
```

```python
plt.figure(figsize=(12,6))

runs_by_over.plot(kind="line", marker="o")

plt.title("Total Runs by Over")
plt.xlabel("Over")
plt.ylabel("Total Runs")

plt.grid()
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

print(numeric_df.columns)
```

```python
correlation = numeric_df.corr()

display(correlation)
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
